### 7. Análisis de población en situación de calle por jurisdicción
Exploramos y analizamos los datos de población en situación de calle por jurisdicción, utilizando un archivo CSV como fuente de datos. El objetivo es identificar las jurisdicciones con mayor porcentaje de población en situación de calle y mostrar esos resultados de manera ordenada.


- Obtener dirección del dataset con la clase Path de pathlib

```python
file_path = pathlib.Path('..','data_modify','c2022_tp_c_resumen_adaptado.csv')
```

>La función `show_homeless_population_jurisdiction` tiene como objetivo mostrar las jurisdicciones con mayor porcentaje de población en situación de calle a partir de un archivo CSV que contiene datos sobre la población en diferentes jurisdicciones.

```python
def homeless_population_jurisdiction (file_path, n):
    """Muestra las jurisdicciones con mayor porcentaje de población en situación de calle.

    Parámetros:
    - n (int): Número de jurisdicciones que se desean mostrar. Por defecto, se muestran las 5 jurisdicciones con mayor porcentaje.

    Retorna:
    - list: Lista de diccionarios donde cada diccionario representa una jurisdicción con su porcentaje de población en situación de calle.
"""
    with open(file_path,'r', encoding='utf-8') as file:
        reader = csv.DictReader(file)
        #lista para almacenar las filas que tienen datos de porcentaje de poblacion en situacion de calle
        homeless_list = []
        
        for row in reader:
            #ignorar la fila 'Total del pais' ya que no es una jurisdiccion
            if row['Jurisdicción'] != 'Total del país':
                #obtener porcentaje
                percentage = row['Porcentaje de poblaci�n en situaci�n de calle']
                #verificar que la celda no esté vacía
                if percentage.strip():
                    #elimina signo % y convierte a numero
                    percentage = float(percentage.rstrip('%'))
                    homeless_list.append({'Jurisdicción': row['Jurisdicción'], 'Porcentaje': percentage})
                
        #ordenar la lista de mayor a menor porcentaje
        homeless_list.sort(key= lambda x: x['Porcentaje'], reverse=True)
        #obtener las primeras n filas con mayor porcentaje
        top_n_jurisdiction = homeless_list[:n]
        
        return top_n_jurisdiction
```

#### Descripción detallada
- **Parámetro 'n'**:La función acepta un parámetro n que indica el número de jurisdicciones que se desean mostrar. Por defecto, se muestran las 5 jurisdicciones con mayor porcentaje.
- **Lectura del Archivo CSV**: Se abre y lee el archivo CSV utilizando la variable file_path. Se utiliza csv.DictReader para manejar las filas del archivo como diccionarios.
- **Creación de la Lista homeless_list**: Se crea una lista vacía homeless_list para almacenar las filas que tienen datos de porcentaje de población en situación de calle.
- **Bucle de Lectura**: Se itera a través de cada fila del archivo. Se excluye la fila 'Total del país' ya que no es una jurisdicción relevante para el análisis.
- **Cálculo y Almacenamiento del Porcentaje**: Se obtiene el porcentaje de población en situación de calle de cada fila y se almacena en la lista homeless_list como un diccionario que contiene la jurisdicción y su respectivo porcentaje.
-**Ordenamiento de la Lista**: Se ordena la lista homeless_list de mayor a menor porcentaje de población en situación de calle.
- **Selección de las Primeras n Jurisdicciones**: Se seleccionan las primeras n filas con mayor porcentaje y se almacenan en la variable top_n_jurisdiction.
- **Retorno de Resultados**: Al finalizar el bucle, se retorna la lista de las primeras n jurisdicciones con mayor porcentaje de población en situación de calle.

> En el siguiente fragmento de código, se le pide al usuario que ingrese la cantidad de jurisdicciones que desea obtener. Si se ingresa un valor válido, se convierte a un entero y se utiliza como la variable n. Si no se ingresa ningún valor o se ingresa un valor no válido, se utiliza el valor predeterminado de 5 para n.

```python
# Leer el valor de n del usuario
n_input = input("Ingrese la cantidad de jurisdicciones que desea obtener (deje en blanco para usar el valor predeterminado de 5): ")
if n_input.strip():  # Verificar si se ingresó algún valor
    try:
        n = int(n_input)
    except ValueError:
        print("Error: Ingrese un valor numérico válido. Se usará el valor predeterminado de 5.")
        n = 5
else:
    n = 5  # Valor predeterminado si no se ingresa ningún valor

homeless_population_jurisdictions = homeless_population_jurisdiction(file_path,n)
for jurisdiction in homeless_population_jurisdictions:
    print(f"{jurisdiction['Jurisdicción']}: {jurisdiction['Porcentaje']}%")
```

Después de leer la cantidad de jurisdicciones desde el usuario y llamar a la función correspondiente, se recorre la lista resultante para mostrar las jurisdicciones con su respectivo porcentaje de población en situación de calle.

### 8. Mayor brecha poblacional
La siguiente función encuentra la jurisdicción donde la brecha entre las poblaciones masculina y femenina es mayor. Utiliza un archivo CSV que contiene datos demográficos de diferentes jurisdicciones, incluyendo la población masculina y femenina.

```python
def largest_gap(file_path):
    """
    Calcula la jurisdicción donde la brecha entre las poblaciones diferenciadas según sexo registrado al nacer es mayor.

    Retorna:
    - str: Nombre de la jurisdicción con mayor brecha poblacional.
    - int: Cantidad de la brecha poblacional entre la población masculina y femenina.
    """
    
    max_gap = 0
    max_jurisdiction = ''
    
    with open(file_path,'r', encoding='utf-8') as file:
        reader = csv.DictReader(file)
        
        for row in reader:
            if row['Jurisdicción'] != 'Total del país':
                men = int(row['Varones Total de población'])
                women = int(row['Mujeres Total de población'])
            
                gap = abs(men - women)
            
                if gap > max_gap:
                    max_gap = gap
                    max_jurisdiction = row['Jurisdicción']
    
    return max_jurisdiction, max_gap
```

#### Descripción detallada de la función
- **Variables iniciales**: Se inicializan las variables max_gap y max_jurisdiction para almacenar la mayor brecha poblacional y la jurisdicción correspondiente respectivamente.
- **Lectura del archivo CSV**: Se abre y lee el archivo CSV con el argumento file_path. Se utiliza csv.DictReader para manejar las filas del archivo como diccionarios.
- **Bucle de lectura**: Se itera a través de cada fila del archivo. Se excluye la fila con 'Jurisdicción' igual a 'Total del país' para evitar cálculos innecesarios.
- **Cálculo de la brecha poblacional**: Se obtienen los valores de población masculina (men) y población femenina (women) de cada fila y se calcula la brecha poblacional como el valor absoluto de la diferencia entre men y women. 
- **Actualización de la mayor brecha**: si la brecha calculada es mayor que *max_gap*, actualiza *max_gap* y *max_jurisdiction* con los valores de la fila actual.
- **Retorno de resultados**: retorna el nombre de la jurisdicción y valor de la brecha.

#### Ejemplo de uso

```python
file_path = pathlib.Path('..', 'data_modify', 'c2022_tp_c_resumen_adaptado.csv')
jurisdiction_name, gap_size = largest_gap(file_path)
print(f"Jurisdicción con mayor brecha poblacional: {jurisdiction_name}")
print(f"Cantidad de brecha poblacional: {gap_size}")
```
