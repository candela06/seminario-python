#### 1. Los diferentes tipos de aeropuertos según la columna 'type'.


>   Importamos los modulos necesarios:

```python
import pathlib
import csv
```

>  Esta función read_csv_file toma la ruta de un archivo CSV como entrada y devuelve dos objetos: un lector CSV y el objeto de archivo correspondiente. Abre el archivo en modo lectura y lo lee con csv.reader(). Luego, salta la primera fila del lector CSV, que generalmente contiene los encabezados.

```python

def read_csv_file(file_path):
    """Lee un archivo CSV y devuelve un objeto lector CSV."""
    file = file_path.open(mode='r', encoding= 'utf-8')
    reader = csv.reader(file)
    header = next(reader)# Leer la primera fila que contiene los encabezados
    return reader, file
```

> Esta función extract_unique_values toma un objeto lector CSV y un índice de columna como entrada y devuelve un conjunto de valores únicos de la columna especificada. Itera sobre cada fila en el lector CSV y agrega el valor de la columna especificada al conjunto list_air_types.


```python

def extract_unique_values(reader,column_index):
    list_air_types = set()
    for row in reader:
       list_air_types.add(row[column_index])
    return list_air_types
```

>Aquí, se define la ruta del archivo CSV que se va a leer. Luego, se llama a la función read_csv_file para obtener el lector CSV y el objeto de archivo correspondiente.

```python
airports_file = pathlib.Path('..','data_modify','ar-airports.csv')    

reader, file = read_csv_file(airports_file)

type_index = 2 # Índice de la columna que quieres acceder

airport_types = extract_unique_values(reader,type_index)

print(airport_types)
file.close()
```

>Se especifica el índice de la columna que quieres acceder en el archivo CSV. Luego, se llama a la función extract_unique_values para obtener los valores únicos de esa columna.
Después de leer el archivo CSV y extraer los valores únicos, se cierra el archivo para liberar los recursos.

#### 2. Los aeropuertos de una elevación, (bajo, medio, alto) cuya especificación se permita modificar fácilmente, evaluando la columna creada 'elevation_name'.


> Para informar los aeropuertos de una elevación específica, es necesario evaluar la columna 'elevation_name' que se ha creado en el archivo de aeropuertos modificados (ar-airports.csv). Esta columna clasifica la elevación de cada aeropuerto en tres categorías: baja, media y alta.

```python
def report_airports_by_elevation(elevation_category, airports_file):
    with open(airports_file, 'r', encoding='utf-8') as file:
        reader = csv.reader(file)
        header = next(reader)  # Leer el encabezado

        elevation_name_index = header.index('elevation_name')

        name_index = header.index('name')


        
        # Filtrar los aeropuertos por la categoría de elevación
        filtered_airports = [[row[name_index]] for row in reader if row[elevation_name_index] == elevation_category]
        
        # Imprimir los aeropuertos encontrados
        print()
        print('-----------------------------------------------')
        print()
        print(f"Aeropuertos de elevación {elevation_category}:")
        print()
        for airport in filtered_airports:
            print(airport)
```

> la función report_airports_by_elevation recibe como parámetros la categoría de elevación deseada ('baja', 'media' o 'alta') y la ruta del archivo de aeropuertos modificados. Luego, utiliza un bucle para filtrar los nombres de los aeropuertos que coinciden con la categoría de elevación especificada y los imprime en la consola.


```python


# Informar los aeropuertos de elevación baja
print()
report_airports_by_elevation('baja', airports_file)
# Informar los aeropuertos de elevación media

report_airports_by_elevation('media', airports_file)

# Informar los aeropuertos de elevación alta
report_airports_by_elevation('alta', airports_file)
```

### 3. Los aeropuertos que tienen una mayor o menor elevación con respecto al valor numérico dado.


> En esta función, elevation_threshold sería el valor numérico dado que se utilizará como umbral de elevación para comparar con los aeropuertos. comparison_type indicaría si quieres encontrar los aeropuertos con una elevación mayor ('mayor') o menor ('menor') que el umbral especificado. Y airports_file sería la ruta del archivo CSV que contiene los datos de los aeropuertos.

```python
def report_airports_by_elevation_difference(elevation_threshold, comparison_type, airports_file):
    with open(airports_file,'r',encoding='utf-8') as file:
        reader = csv.reader(file)
        header = next(reader)
        
        elevation_index = header.index('elevation_ft')
        name_index = header.index('name')
        if comparison_type == 'mayor':
            filtered_airports = [[row[name_index]] for row in reader if row[elevation_index] and (str(row[elevation_index]).strip().isdigit() and int(row[elevation_index]) > elevation_threshold)]
        else:
            filtered_airports = [[row[name_index]] for row in reader if row[elevation_index] and (str(row[elevation_index]).strip().isdigit() and int(row[elevation_index]) < elevation_threshold)]

    # Imprimir los aeropuertos encontrados
    print()
    print('-----------------------------------------------')
    print()
    print(f"Aeropuertos de elevación {comparison_type} a {elevation_threshold}: ")
    print()
    for airport in filtered_airports:
        print(airport)
```

> Utilizamos elevation_index = header.index('elevation_ft') y name_index = header.index('name') para obtener los índices de las columnas 'elevation_ft' y 'name' respectivamente. Estos índices nos permitirán acceder a los valores de estas columnas en cada fila.


> Filtrado de aeropuertos: Utilizamos comprensiones de lista para filtrar las filas del archivo CSV según ciertas condiciones. Para el filtrado, comprobamos si la columna de elevación contiene un valor y si es un número mediante la expresión (str(row[elevation_index]).strip().isdigit(). Si cumple esta condición, convertimos el valor de la elevación a un entero con int(row[elevation_index]) y lo comparamos con el umbral dado (elevation_threshold). Dependiendo del tipo de comparación (comparison_type), seleccionamos aeropuertos cuya elevación sea mayor o menor que el umbral.

>  Imprimimos los aeropuertos encontrados para la categoría de elevación dada. Iteramos sobre la lista filtered_airports y mostramos el nombre de cada aeropuerto.

```python
num = input('Ingresar un valor numerico para comparar si su elevacion es menor o mayor: ')

report_airports_by_elevation_difference(int(num),'mayor',airports_file)

report_airports_by_elevation_difference(int(num),'menor',airports_file) 
```

> Finalmente, llamamos a la función report_airports_by_elevation_difference dos veces, una para encontrar los aeropuertos con elevación mayor que el umbral y otra para encontrar los aeropuertos con elevación menor que el umbral. En cada llamada, pasamos el umbral deseado y el tipo de comparación ('mayor' o 'menor').
