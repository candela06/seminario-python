#### 🛜 10. Mostrar las cantidades de localidades con cada tipo de conectividad.

>La funcion `connectivity_by_city` analiza el archivo CSV que contiene información sobre la conectividad de ciudades, iterando sobre el mismo. Determina si una ciudad tiene un tipo específico de conectividad examinando las columnas correspondientes y contabilizando el número de ciudades para cada tipo de conectividad. Al finalizar, devuelve un diccionario con la cantidad de ciudades que tiene cada tipo de conectividad.

```python
def connectivity_by_city(path):
    connectivity = {}
    with open(path, 'r',encoding='utf8',newline='') as f:
        reader = csv.DictReader(f)
        for row in reader:
            for types_of_connectivity in list(row.keys())[4:13]:
                if row[types_of_connectivity] == 'SI':
                    if types_of_connectivity not in connectivity: connectivity[types_of_connectivity] = 1
                    else:connectivity[types_of_connectivity] += 1
    return connectivity
```

>Implementación

```python
with open(path, 'r') as file:
    connectivity = connectivity_by_city(path)
    for connectivity_type, count in connectivity.items():
        print(f"La cantidad de localidades con {connectivity_type} son: {count}")
```

#### 🛜 11. Mostrar las provincias para las cuales todas sus ciudades poseen Fibra Óptica

> La función `connectivity_by_province` analiza el archivo CSV que contiene información sobre la conectividad de provincias. Itera sobre el archivo para determinar que provincias poseen conectividad de fibra óptica en todas sus localidades. Luego, retorna una lista de las provincias que cumplen con lo dicho.

```python
def connectivity_by_province(path):
    provinces = []
    with open(path, 'r',encoding='utf8',newline='') as f:
        reader = csv.DictReader(f)
        for row in reader:
            province = row['Provincia']
            fiber_optic = True
            for row2 in reader:
                if row2['Provincia'] == province and row2['FIBRAOPTICA'] == 'NO':
                    fiber_optic = False
                    break
            if fiber_optic: provinces.append(province)
    return provinces
```

>Implementación

```python
provinces = connectivity_by_province(path)
print (f"Las provincias que tienen fibra óptica en todas sus localidades son:", ",".join(provinces))
```

#### 🛜 12. Mostrar para cada provincia su capital y, si se conoce la información para dicha capital, informar si posee conectividad.  En caso de no conocer la información, mostrar el texto “conectividad desconocida”.

La función `get_province_capitals` tiene como objetivo saber cuales son las capitales de cada provincia. Analizando el archivo CSV, itera sobre el mismo para guardar las capitales en un diccionario junto con sus respectivas provincias. Luego, retorna el mismo.

```python
def get_province_capitals(path_capitals):
    capitals = {}
    with open(path_capitals, 'r', encoding='utf-8') as file:
        reader = csv.DictReader(file)
        for row in reader:
            city = unidecode (row['city']).upper()
            province = unidecode(row['admin_name']).upper()
            if row['capital'] == 'admin': capitals[province] = city
    return capitals
```

> La función `connectivity_by_capital` analiza un archivo CSV que contiene información sobre la conectividad de localidades y devuelve un diccionario que indica si cada capital tiene conectividad. Utiliza el diccionario `capitals` creado anteriormente que mapea los nombres de capitales a sus valores correspondientes para determinar que localidades son capitales de la provincia.



```python
def connectivity_by_capital(path, capitals):
    connectivity = {}
    with open(path,'r',encoding='utf8',newline='') as file:
        reader = csv.DictReader(file)
        for row in reader:
            locality = unidecode(row['Localidad']).upper() 
            if locality in capitals.values(): 
                has_connectivity = row['posee_conectividad'] == 'SI'
                connectivity[locality] = has_connectivity
    return connectivity
```

>Implementación

```python
capitals = get_province_capitals(path_capitals)
connectivity_capitals = connectivity_by_capital(path, capitals)
for province, capital in capitals.items():
    if capital in connectivity_capitals: has_connectivity = "posee conectividad" if connectivity_capitals[capital] else "no posee conectividad"
    else: has_connectivity = "conectividad desconocida"
    print(f"La capital de {province} es {capital} y {has_connectivity}")
```
